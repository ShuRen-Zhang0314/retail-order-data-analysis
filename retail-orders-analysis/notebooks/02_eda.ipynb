```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
os.makedirs('../visualizations', exist_ok=True)

plt.rcParams['font.sans-serif'] = ['SimHei']  # 中文显示
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_csv('../data/orders_cleaned.csv', parse_dates=['order_date'])
df.describe()   # 数值字段的均值/分位数/极值

import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.sans-serif'] = ['SimHei']  # Windows 中文显示
plt.rcParams['axes.unicode_minus'] = False

monthly = df.groupby(df['order_date'].dt.to_period('M'))['sale_price'].sum()
monthly.plot(kind='line', figsize=(12, 5), title='月度销售额趋势')
plt.ylabel('销售额（¥）')
plt.savefig('../visualizations/monthly_trend.png', dpi=150, bbox_inches='tight')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df.groupby('category')['sale_price'].sum().plot(kind='bar', ax=axes[0], title='各品类销售额')
df.groupby('region')['profit'].sum().plot(kind='bar', ax=axes[1], title='各地区利润')
plt.savefig('../visualizations/category_region.png', dpi=150, bbox_inches='tight')

sns.scatterplot(data=df, x='discount_percent', y='profit', alpha=0.3)
plt.title('折扣率 vs 利润')
plt.savefig('../visualizations/discount_vs_profit.png', dpi=150, bbox_inches='tight')

sns.heatmap(df[['quantity','discount','sale_price','profit']].corr(),
            annot=True, cmap='RdYlGn', center=0)
plt.title('数值字段相关性')
plt.savefig('../visualizations/correlation.png', dpi=150, bbox_inches='tight')
```
